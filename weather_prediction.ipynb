```python
# ==============================
# Weather Data Analysis & Prediction (Google Colab)
# ==============================

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

print("PROGRAM STARTED")

# Load dataset
df = pd.read_csv("weather.csv")
print("CSV LOADED")

# Show first 5 rows
print("First 5 rows of dataset:")
display(df.head())

# Select required columns
df = df[['Year', 'Month', 'Day', 'AvgTemperature']]
print("COLUMNS SELECTED")

# Remove missing values
df = df.dropna()
print("MISSING VALUES REMOVED")

# Create time index
df['TimeIndex'] = range(len(df))
print("TIME INDEX CREATED")

# Plot temperature trend
plt.figure(figsize=(10,5))
plt.plot(df['TimeIndex'], df['AvgTemperature'])
plt.xlabel("Time")
plt.ylabel("Average Temperature")
plt.title("Temperature Trend Over Time")
plt.show()

# Prepare data for model
X = df[['TimeIndex']]
y = df['AvgTemperature']

# Train Linear Regression model
model = LinearRegression()
model.fit(X, y)
print("MODEL TRAINED")

# Predict future temperature
future_days = 30
future_time = [[len(df) + future_days]]
prediction = model.predict(future_time)

print(f"Predicted Average Temperature after {future_days} days:", prediction[0])
```
